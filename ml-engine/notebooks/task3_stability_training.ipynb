## Data files
Upload `task3_train_filtered_annotations.csv` and `task3_val_filtered_annotations.csv` once to `MyDrive/Synapse/Task3/data/`. Do not upload the test split.


```python
from pathlib import Path
root = Path("/content/drive/MyDrive/Synapse/Task3")
train = root / "data/task3_train_filtered_annotations.csv"
val = root / "data/task3_val_filtered_annotations.csv"
output = root / "artifacts/task3_prototype_v1"
for f in (train, val):
    assert f.is_file(), f"Missing Google Drive dataset: {f}"
output.mkdir(parents=True, exist_ok=True)
print("Input CSVs found; artifacts will be saved to:", output)
```

```python
import subprocess, sys
script = repo / "ml-engine/scripts/train_stability.py"
assert script.is_file(), f"Training script not found: {script}. Push the script to GitHub first."
subprocess.run([sys.executable, "-m", "pip", "install", "-q", "-e", str(repo / "ml-engine")], check=True)
subprocess.run([sys.executable, str(script), "--train", str(train), "--val", str(val), "--output", str(output)], check=True)
```
